# 概要

課題に取り組む前の状態では、以下の2つのコードセルが用意されています。

```python
#@title 九九のリストを作成する

```
```python
#@title 九九の表を表示する
for v in range(1, 10):
    for h in range(1, 10):
        print(f"{(v * h):3d}", end="")
    
    print("")
```

2つ目のセル（`九九の表を表示する`）を実行すると、以下のような出力結果になります。
```
  1  2  3  4  5  6  7  8  9
  2  4  6  8 10 12 14 16 18
  3  6  9 12 15 18 21 24 27
  4  8 12 16 20 24 28 32 36
  5 10 15 20 25 30 35 40 45
  6 12 18 24 30 36 42 48 54
  7 14 21 28 35 42 49 56 63
  8 16 24 32 40 48 56 64 72
  9 18 27 36 45 54 63 72 81
```

現状のコードについて補足します。
- `{(v * h):3d}`の`3d`は、書式指定文字列となります。`3d` は「3桁の整数値として表示する（3桁未満の場合はスペースで埋めて疑似的に3文字にする）」という書式です。
- print() の引数の end="" は、print() したあとで改行されなくしています。print() は通常、出力内容の末尾で自動的に改行するためです。九九の表として、数字を横並びで表示させたいため、end="" を指定しています。
- ただし9の段（9, 18, 27, ...）に来たら改行したいため、内側の for 文の処理が終わったら print("") を実行し、改行するようにしています。

- `#@title 九九の表を表示する` と書かれた行の下にある二重の for 文を修正してください：

  - その際、前述の実行結果の表示と同じレイアウトにしてください


````python
#@title 九九のリストを作成する
l99 = [[v*h for h in range(1, 10)] for v in range(1, 10)]
````
